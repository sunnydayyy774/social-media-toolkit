## 3. 视频层情感分析

### 3.1 分析单位与输入文本

视频层情感分析的单位是单个视频帖子。

输入文本使用：

`text_clean_bertopic_with_asr`

也就是将视频标题/描述与 ASR 音频转写文本合并后的清洗文本。

使用模型：

`lxyuan/distilbert-base-multilingual-cased-sentiments-student`

情感标签包括：

- `negative`
- `neutral`
- `positive`

注意：这里的情感标签不等同于立场标签。`positive` 不代表支持日本修宪，`negative` 也不直接等于反对日本修宪。

```python
post_success = posts_sentiment[posts_sentiment['sentiment_status'].eq('success')].copy()
post_counts = post_success['sentiment_label'].value_counts().reindex(['negative', 'neutral', 'positive'], fill_value=0)
post_ratio = (post_counts / post_counts.sum() * 100).round(2)
pd.DataFrame({'post_count': post_counts, 'ratio_percent': post_ratio})
```

### 3.2 视频情感总体分布

视频情感分析成功处理 3,344 条视频文本。其中：

- negative：2,523 条，占 75.45%
- neutral：98 条，占 2.93%
- positive：723 条，占 21.62%

```python
show_image(POST_SENTIMENT_VIZ / 'sentiment_distribution.png', width=900)
```

### 3.3 视频情感随时间变化

这张图展示不同情感类别的视频数量如何随发布时间变化。

```python
show_image(POST_SENTIMENT_VIZ / 'sentiment_over_time_year_count.png', width=1000)
```

观察：视频数量在后期明显增加，negative 视频数量也随之上升。这说明相关讨论升温时，视频文本中的负面情绪表达也同步增加。

### 3.4 Topic × 视频情感比例

这张热力图展示不同 topic 内部的视频情感比例。

横轴是情感类别，纵轴是 topic，颜色表示该 topic 内部对应情感的比例。

```python
show_image(POST_SENTIMENT_VIZ / 'topic_sentiment_heatmap_ratio.png', width=900)
```

观察：不同 topic 的视频情感分布存在差异。部分 topic 中 negative 比例更高，说明这些主题更容易承载警惕、批评或担忧等情绪表达。

### 3.5 高互动视频情感分布

这张图选取评论数最高的一批视频，观察这些高互动视频本身的情感分布。

```python
show_image(POST_SENTIMENT_VIZ / 'top_interaction_sentiment.png', width=800)
```

### 3.6 视频情感分析小结

视频层分析说明：

- 视频文本整体以 negative 为主。
- 讨论升温的后期，negative 视频数量也明显增加。
- 高互动视频和不同 topic 之间存在情感差异。
- 但情感分析只能说明表达情绪，不等于支持/反对日本修宪的立场分类。

## 4. 全量评论层情感分析

### 4.1 分析单位与数据规模

评论层情感分析的单位是单条评论。

本次对 `include_in_analysis == true` 的评论进行全量情感分析。

结果规模：

- 成功分析评论：1,135,397 条
- 空评论：0 条
- 可用于视频情感与评论情感对比的评论：1,135,141 条

少掉的 256 条不是评论情感缺失，而是对应视频缺少有效的视频情感标签，因此不能进入“视频情感 vs 评论情感”的对比图。

```python
comment_success = comments_sentiment[comments_sentiment['sentiment_status'].eq('success')].copy()
comment_counts = comment_success['comment_sentiment_label'].value_counts().reindex(['negative', 'neutral', 'positive'], fill_value=0)
comment_ratio = (comment_counts / comment_counts.sum() * 100).round(2)
pd.DataFrame({'comment_count': comment_counts, 'ratio_percent': comment_ratio})
```

### 4.2 评论情感总体分布

全量评论情感结果为：

- negative：544,784 条，占 47.98%
- neutral：86,394 条，占 7.61%
- positive：504,219 条，占 44.41%

评论层面不是单纯一边倒负面，而是 negative 与 positive 都较多，neutral 明显较少。

```python
show_image(COMMENT_SENTIMENT_VIZ / 'comment_sentiment_distribution_full.png', width=900)
```

### 4.3 评论情感随时间变化

这张图展示评论情感数量随评论发布时间年份的变化。

```python
show_image(COMMENT_SENTIMENT_VIZ / 'comment_sentiment_over_time_year_count_full.png', width=1000)
```

观察：评论数量在后期明显增加，negative 和 positive 评论都大量增长。这说明评论区表达更活跃，但情感表达并不只表现为单一负面。

### 4.4 视频情感 vs 评论情感矩阵

这张矩阵热力图用于比较视频本身的情感和评论区情感是否一致。

横轴是视频情感，纵轴是评论情感，颜色越深表示评论数量越多。

```python
show_image(COMMENT_SENTIMENT_VIZ / 'post_comment_sentiment_matrix_full.png', width=850)
```

观察：视频情感和评论情感并不完全一致。评论区可能对视频内容进行赞同、反驳、嘲讽或延伸讨论，因此评论情感不能简单由视频情感推出。

### 4.5 不同视频情感下的负面评论比例

这张图按视频情感分组，统计每组视频下面评论中 negative 的比例。

```python
show_image(COMMENT_SENTIMENT_VIZ / 'comment_negative_ratio_by_post_sentiment_full.png', width=850)
```

结果显示：

- negative 视频下，negative 评论比例为 46.1%。
- neutral 视频下，negative 评论比例为 53.1%。
- positive 视频下，negative 评论比例为 55.3%。

这说明即使视频文本被模型判为 neutral 或 positive，评论区仍然可能出现较高比例的 negative 评论。

### 4.6 Topic × 年份负面评论比例

这张散点图中每个点代表一个评论用户。

### 4.8 评论情感分析小结

评论层分析说明：

- 全量评论中 negative 最多，但 positive 也非常接近，neutral 较少。
- 评论区情感和视频文本情感不完全一致。
- 即使视频本身被判为 neutral 或 positive，评论区仍可能出现较高比例的 negative 评论。
- 情感分析不等同于立场分析；后续如果要回答“是否反对日本修宪”，需要进一步做 stance analysis。

## 5. 当前局限与下一步

### 5.1 当前局限

1. 情感分析不是立场分析，不能直接解释为支持或反对日本修宪。
2. 普通情感模型可能误判讽刺、反问、政治口号和语境化表达。
3. `positive` 可能来自“和平”“胜利”“说得好”等词，不代表支持日本修宪。
4. 评论情感和视频情感的差异需要结合具体文本样本进一步解释。

### 5.2 下一步计划

1. 对视频和评论情感结果进行人工抽样检查。
2. 继续推进立场分析，将标签改为“反对日本修宪 / 支持日本修宪 / 中立陈述 / 不明确”。
3. 结合 topic、时间和评论网络，分析哪些主题在 2025-2026 年更容易引发强烈回应。
